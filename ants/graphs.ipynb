# This file contains matplotlib graphs for visualising different ant colony statistics

```python
from algorithm import AntColony
import matplotlib.pyplot as plt
import random 
from helper import calculate_distances
import numpy as np

SCREENW = 1000
SCREENH = 800
NUM_CITIES = 100
ITERATIONS = 200

cities = []
for _ in range(NUM_CITIES):
    x = random.randint(0, SCREENW - 1)
    y = random.randint(0, SCREENH - 1)
    cities.append((x, y))

distances = calculate_distances(cities)
ant_colony = AntColony(distances, n_ants=10)
```

```python
x = []
y = []

for i in range(ITERATIONS):
    best_path, best_path_length, pheromones = ant_colony.run()
    x.append(i)
    y.append(best_path_length)

plt.plot(x, y)
plt.show()  # The colony should converge on the minimal path length
```

```python
ant_colony.reset_pheromones()
x = []
y = []
y2 = []
y3 = []

for i in range(ITERATIONS):
    best_path, best_path_length, pheromones = ant_colony.run()
    x.append(i)
    y.append(np.amax(pheromones))
    y2.append(np.amin(pheromones))
    y3.append(np.average(pheromones))

g1 = plt.plot(x, y)
g2 = plt.plot(x, y2, '--')
g3 = plt.plot(x, y3, '--')
plt.yscale('log',base=2)
plt.show()
```
